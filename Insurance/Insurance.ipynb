```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv(r"C:\Users\VARSHINI\OneDrive - Rathinam Group Of Institutions\Desktop\Deep_Learning\Insurance\insurance_data.csv")
```

```python
print(df.head())
```

```python
print(df.shape)
print(df.info())
print(df.describe())
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['age']],df.bought_insurance,train_size=0.8)
```

```python
log_model = LogisticRegression()
```

```python
log_model.fit(X_train, y_train)
```

```python
y_predicted = log_model.predict(X_test)
```

```python
log_model.predict_proba(X_test)
```

```python
y_predicted
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_predicted)

print("Accuracy =", accuracy)
```

```python
log_model.predict([[35]])
```

```python
log_model.predict_proba([[35]])
```

```python
m = log_model.coef_[0][0]
c = log_model.intercept_[0]
```

```python
print("Slope =", m)
print("Intercept =", c)
```

```python
x = 35
z = (m * x) + c
print("z =", z)
```

```python
sigmoid = 1 / (1 + np.exp(-z))
```

```python
print("Sigmoid Output =", sigmoid)
```

```python
# Create age values
ages = np.linspace(df.age.min(), df.age.max(), 300).reshape(-1,1)

# Predict probabilities
probabilities = log_model.predict_proba(ages)[:,1]

# Plot graph
plt.figure(figsize=(8,5))

# Actual data points
plt.scatter(df['age'], df['bought_insurance'], color='blue')

# Sigmoid curve
plt.plot(ages, probabilities, color='red')

plt.xlabel('Age')
plt.ylabel('Probability of Buying Insurance')
plt.title('Logistic Regression Sigmoid Curve')

plt.grid(True)
plt.show()
```
